# Topic 1 – What is Inference?

**Inference** is the process of using a trained machine learning model to generate predictions for new input data.

A model learns patterns during training. After training, we provide new input data, and the model produces an output.

**Example:**

A sentiment analysis model receives a sentence and predicts whether the sentiment is Positive or Negative.

**Workflow:**

Input Text  
↓  
Trained Model  
↓  
Prediction  
↓  
Output (Positive / Negative)

**Important:**

- Training means teaching the model using data.
- Inference means using the trained model to make predictions.
- Batch inference means providing multiple inputs together to the model.

```python
from transformers import pipeline

# Load pre-trained sentiment model
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

# Single input
result = classifier("I received salary")

print(result)
```

# Topic 2 – Single Inference

**Single inference** means providing one input at a time to a trained model and receiving one prediction.

The model processes each input separately.

**Example:**

Suppose we have three customer reviews:

- Review 1: I love this product.
- Review 2: The service is very bad.
- Review 3: The experience was excellent.

In single inference, each review is passed to the model separately.

**Workflow:**

Input 1 → Model → Output 1

Input 2 → Model → Output 2

Input 3 → Model → Output 3

**Characteristics:**

- Processes one input at a time.
- Simple to implement.
- Useful for real-time predictions.
- Multiple inputs require multiple inference calls.

```python
from transformers import pipeline
import time

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent"
]

start_time = time.time()

# Single inference
for text in texts:
    result = classifier(text)
    print(text, ":", result)

end_time = time.time()

execution_time = end_time - start_time

print("Total execution time:", execution_time)
```

# Topic 3 – Batch Inference

**Batch inference** means providing multiple inputs together to a trained model and receiving multiple predictions in one inference call.

Instead of processing each input separately, we collect multiple inputs into a batch.

**Example:**

Suppose we have three customer reviews:

- Review 1: I love this product.
- Review 2: The service is very bad.
- Review 3: The experience was excellent.

In batch inference, all three reviews are sent to the model together.

**Workflow:**

Inputs (1, 2, 3)
↓
Batch
↓
Trained Model
↓
Outputs (1, 2, 3)

**Characteristics:**

- Processes multiple inputs together.
- Reduces repeated inference calls.
- Can improve processing efficiency.
- Requires more memory as batch size increases.
- Useful for processing large amounts of data.

```python
from transformers import pipeline
import time

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent"
]

start_time = time.time()

# Batch inference
results = classifier(texts, batch_size=3)

for text, result in zip(texts, results):
    print(text, ":", result)

end_time = time.time()

execution_time = end_time - start_time

print("Total execution time:", execution_time)
```

# Topic 4 – Single Inference vs Batch Inference Comparison

Single inference and batch inference are two ways of processing input data using a trained model.

**1. Single Inference**

- Processes one input at a time.
- Each input requires a separate model call.
- Suitable for individual or real-time predictions.
- Repeated calls may introduce additional overhead.

**2. Batch Inference**

- Processes multiple inputs together.
- Multiple inputs are passed in one model call.
- Can improve processing efficiency.
- Requires more memory for larger batches.

**Comparison:**

| Feature | Single Inference | Batch Inference |
|---|---|---|
| Input | One at a time | Multiple together |
| Model calls | Multiple | Fewer |
| Processing | Individual | Grouped |
| Memory usage | Usually lower | Usually higher |
| Throughput | Often lower | Can be higher |
| Use case | Real-time prediction | Bulk prediction |

**Conclusion:**

Single inference is useful when predictions are needed individually.

Batch inference is useful when many predictions need to be processed efficiently.

```python
import time
from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent",
    "The quality is amazing",
    "I am disappointed"
]

# Single inference
start = time.time()

single_results = []

for text in texts:
    single_results.extend(classifier(text))

single_time = time.time() - start


# Batch inference
start = time.time()

batch_results = classifier(texts, batch_size=5)

batch_time = time.time() - start


# Comparison
print("Single inference time:", single_time)
print("Batch inference time:", batch_time)

print("\nSingle predictions:", single_results)
print("\nBatch predictions:", batch_results)
```

# Topic 4 – Throughput Measurement

**Throughput** is the number of inputs a model can process within one second.

It helps us understand how efficiently a model handles multiple inputs.

In batch inference, throughput is useful for measuring how many predictions the model can generate per second.

**Formula:**

Throughput = Total Number of Inputs / Total Execution Time

**Example:**

Suppose a model processes 100 text inputs in 5 seconds.

Throughput = 100 / 5 = 20 inputs per second.

**Comparison:**

- Single inference processes inputs individually.
- Batch inference processes multiple inputs together.
- Higher throughput means more inputs are processed per second.

**Why measure throughput?**

- To compare single and batch inference.
- To understand model processing capacity.
- To evaluate inference efficiency.
- To identify a suitable batch size.

```python
import time
from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    device=-1
)

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent",
    "The quality is amazing",
    "I am disappointed"
]

# Warm-up
classifier(texts[:1])

# Single inference
start = time.perf_counter()

for text in texts:
    classifier(text)

single_time = time.perf_counter() - start

# Batch inference
start = time.perf_counter()

classifier(texts, batch_size=5)

batch_time = time.perf_counter() - start

# Calculate throughput
single_throughput = len(texts) / single_time
batch_throughput = len(texts) / batch_time

print("Single throughput:", round(single_throughput, 2), "inputs/second")
print("Batch throughput:", round(batch_throughput, 2), "inputs/second")
```

# Topic 5 – Memory Usage Measurement

**Memory usage** refers to the amount of RAM consumed by a model while processing input data.

During inference, the model requires memory to store:

- Model parameters (weights).
- Input data.
- Intermediate calculations.
- Model outputs.

In batch inference, multiple inputs are processed together. Therefore, memory consumption can increase as the batch size increases.

**Why measure memory usage?**

- To understand how much RAM the model requires.
- To compare single and batch inference.
- To identify memory limitations.
- To select a suitable batch size.

**Comparison:**

- Single inference generally requires less temporary memory.
- Batch inference may require more temporary memory.
- Larger batches can increase memory consumption.

**Important:** Model memory is also required in both approaches. The memory used for model weights does not disappear between individual predictions.

```python
import psutil
import os
import gc
from transformers import pipeline

process = psutil.Process(os.getpid())

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    device=-1
)

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent",
    "The quality is amazing",
    "I am disappointed"
]

# Warm-up
classifier(texts[:1])

# Single inference memory
gc.collect()

before_single = process.memory_info().rss / (1024 ** 2)

for text in texts:
    classifier(text)

after_single = process.memory_info().rss / (1024 ** 2)

single_memory = after_single - before_single


# Batch inference memory
gc.collect()

before_batch = process.memory_info().rss / (1024 ** 2)

classifier(texts, batch_size=5)

after_batch = process.memory_info().rss / (1024 ** 2)

batch_memory = after_batch - before_batch


print("Single inference memory change:",
      round(single_memory, 2), "MB")

print("Batch inference memory change:",
      round(batch_memory, 2), "MB")
```

# Topic 6 – Batch Size vs Resource Consumption

**Batch size** is the number of input samples processed together during one model inference operation.

For example:

- Batch size 1 means one input at a time.
- Batch size 2 means two inputs together.
- Batch size 5 means five inputs together.
- Batch size 10 means ten inputs together.

**How does batch size affect performance?**

When batch size increases:

- More inputs are processed together.
- Throughput may increase.
- Memory consumption generally increases.
- Processing can become slower if available resources are insufficient.

When batch size decreases:

- Memory consumption is generally lower.
- More inference calls may be required.
- Throughput may decrease.

**Trade-off:**

| Batch Size | Throughput | Memory Usage |
|---|---|---|
| Small | May be lower | Generally lower |
| Medium | May improve | Moderate |
| Large | May improve further | Generally higher |

**Why is batch size important in enterprise environments?**

Enterprise applications often process thousands or millions of records.

Choosing a suitable batch size helps balance:

- Processing speed.
- RAM consumption.
- CPU/GPU resources.
- System stability.

**Conclusion:**

A larger batch size is not always better. We should choose a batch size that provides good throughput without exceeding available system resources.

```python
import time
import psutil
import os
import gc

from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    device=-1
)

process = psutil.Process(os.getpid())

texts = [
    "I love this product",
    "The service is very bad",
    "The experience was excellent",
    "The quality is amazing",
    "I am disappointed",
    "This product is wonderful",
    "The delivery was very slow",
    "I am happy with the service"
]

batch_sizes = [1, 2, 4, 8]

results = []

# Warm-up
classifier(texts[:1])

for size in batch_sizes:

    gc.collect()

    memory_before = process.memory_info().rss / (1024 ** 2)

    start = time.perf_counter()

    predictions = classifier(texts, batch_size=size)

    execution_time = time.perf_counter() - start

    memory_after = process.memory_info().rss / (1024 ** 2)

    memory_change = memory_after - memory_before

    throughput = len(texts) / execution_time

    results.append({
        "Batch Size": size,
        "Execution Time": round(execution_time, 4),
        "Throughput": round(throughput, 2),
        "Memory Change (MB)": round(memory_change, 2)
    })

import pandas as pd

df = pd.DataFrame(results)

print(df)
```
