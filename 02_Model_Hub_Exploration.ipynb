## Topic 1: Searching the Model Hub

**Concept**
The Hugging Face Hub hosts hundreds of thousands of models. You can search it on the website or from Python using `HfApi().list_models()`. You can filter by task, language, library and license, and sort by downloads or likes.

**Why it's needed**
Picking the first model you see is risky. A search with filters shows you the real options, and sorting by downloads shows which models the community trusts.

```python
from huggingface_hub import HfApi

api = HfApi()

try:
    models = api.list_models(
        task="text-classification",
        language="en",
        sort="downloads",
        limit=5,
    )
    for m in models:
        print(m.id, "| downloads:", m.downloads, "| likes:", m.likes)
except Exception as e:
    print("Could not reach the Hub (check internet):", type(e).__name__)
```

**What this example tells us**
One short call returns the five most-downloaded English text-classification models. Changing `task`, `language` or `sort` changes the shortlist. A shortlist is the starting point for comparison, not the final answer.

**AI/ML Example**
Before fine-tuning a sentiment model, search for `text-classification` models in your language. An existing fine-tuned model may already solve the task and save days of training.

**Business Example**
A support team wants to tag customer emails by sentiment. A 2-minute Hub search shows ready-made models, so the team can test a prototype the same day.

**AI/ML Engineer Use Case**
Wrap the search in a helper function so any teammate can shortlist candidate models for a new task using the same filters and the same sort order.

## Topic 3: Model Comparison Table

**Concept**
A comparison table puts the key fields of all candidate models side by side: Model, Task, Architecture, Parameters, Size, Language and License.

**Why it's needed**
One table lets you see trade-offs at a glance, for example a small model that runs on a laptop versus a large model that needs a GPU.

## Topic 4: Selecting a Model by Requirements

**Concept**
Instead of picking the first result, write down the requirements (task, language, maximum size, allowed license) and filter the candidates against them. Then rank what is left.

**Why it's needed**
Requirements make the decision repeatable and explainable. If the requirements change, you re-run the filter and get a new answer.

```python
ALLOWED_LICENSES = {"Apache-2.0", "MIT"}   # licenses our company accepts

def select_model(task_keyword, language, max_size_mb):
    results = []
    for m in models:
        task_ok = task_keyword.lower() in m["Task"].lower()
        lang_ok = m["lang"] in (language, "multi")
        size_ok = m["size_mb"] <= max_size_mb
        lic_ok = m["License"] in ALLOWED_LICENSES
        if task_ok and lang_ok and size_ok and lic_ok:
            results.append(m)
    results.sort(key=lambda m: m["size_mb"])      # smallest first
    return results

scenarios = [
    ("Scenario 1: English sentiment, laptop CPU, under 500 MB", "text classification", "en", 500),
    ("Scenario 2: speech-to-text for Hindi, under 1500 MB", "speech", "hi", 1500),
    ("Scenario 3: English summaries, under 500 MB", "summarization", "en", 500),
]

for title, task, lang, size in scenarios:
    print(title)
    found = select_model(task, lang, size)
    if found:
        for m in found:
            print("   ->", m["Model name"], "|", m["Size"], "|", m["License"])
    else:
        print("   -> No model meets all requirements. Relax a requirement or search again.")
    print()
```

**What this example tells us**
The same function gives three different answers because the requirements differ. Scenario 3 returns nothing: the only summarization model in our list is too large. That is a useful result, because it tells you to search the Hub for a smaller summarizer instead of forcing a bad fit.

**AI/ML Example**
For Hindi speech, a multilingual model such as Whisper passes the language check, while English-only models are removed automatically.

**Business Example**
A product manager can say "English, CPU only, under 500 MB, commercial license", and the team can show exactly why a model was chosen or rejected.

**AI/ML Engineer Use Case**
Extend the filter with more fields (latency, accuracy on your own test set, hardware). The final step is always to test the top 2 or 3 models on your own data before deciding.
