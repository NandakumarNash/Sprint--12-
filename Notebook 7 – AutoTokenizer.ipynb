1. What is AutoTokenizer?

AutoTokenizer is a Hugging Face class used to automatically load the correct tokenizer for a pretrained model.

Different models may use different tokenization methods.

Instead of manually selecting the tokenizer class, we can use AutoTokenizer.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)
```

```python
Here:
- AutoTokenizer – Automatically selects the correct tokenizer.
- from_pretrained() – Loads a pretrained tokenizer.
- "bert-base-uncased" – Hugging Face model repository name.
- tokenizer – Stores the loaded tokenizer.
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

print(tokenizer)
```

```python
3. Encoding Text
Encoding means converting normal text into numerical token IDs that a model can understand.
```

```python
text = "I am learning Hugging Face"

encoded = tokenizer(text)

print(encoded)
```

```python
The output contains fields such as:
- input_ids
- token_type_ids
- attention_mask
```

```python
4. Decoding Tokens
Decoding converts token IDs back into readable text.
```

```python
text = "I am learning Hugging Face"

encoded = tokenizer(text)

decoded = tokenizer.decode(
    encoded["input_ids"]
)

print(decoded)
```

```python
Understand:
- Encoding: Text → Token IDs
- Decoding: Token IDs → Text
Note: Special tokens may appear in decoded text. Use skip_special_tokens=True to remove them.
```

```python
decoded = tokenizer.decode(
    encoded["input_ids"],
    skip_special_tokens=True
)

print(decoded)
```

```python
5. Batch Encoding
Batch encoding means processing multiple text inputs together instead of one at a time.
```

```python
texts = [
    "I am learning Python",
    "I am learning Deep Learning",
    "I am learning Hugging Face"
]

encoded = tokenizer(
    texts,
    padding=True,
    truncation=True
)

print(encoded["input_ids"])
print(encoded["attention_mask"])
```

```python
Understand:
- Single encoding: One text at a time.
- Batch encoding: Multiple texts together.
```

```python
6. Padding
Padding adds special padding tokens to shorter sequences so that all sequences in a batch have equal lengths.
```

```python
texts = [
    "I like AI",
    "I am learning Artificial Intelligence"
]

encoded = tokenizer(
    texts,
    padding=True
)

print(encoded["input_ids"])
print(encoded["attention_mask"])
```

```python
7. Truncation
Truncation removes extra tokens when the input text exceeds the specified maximum sequence length.
```

```python
text = "I am learning Natural Language Processing using Hugging Face Transformers"

encoded = tokenizer(
    text,
    truncation=True,
    max_length=8
)

print(encoded["input_ids"])
```

Understand:
- truncation=True – Enables truncation.
- max_length=8 – Limits the sequence to 8 tokens, including special tokens.

```python
8. Attention Masks
An attention mask tells the model which tokens are actual input tokens and which are padding tokens.
- 1 – Actual token.
- 0 – Padding token.
```

```python
texts = [
    "I like AI",
    "I am learning Artificial Intelligence"
]

encoded = tokenizer(
    texts,
    padding=True
)

print("Input IDs:")
print(encoded["input_ids"])

print("Attention Mask:")
print(encoded["attention_mask"])
```

```python
from transformers import AutoTokenizer, AutoModel

# Step 1: Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

# Step 2: Load pretrained model
model = AutoModel.from_pretrained(
    "bert-base-uncased"
)

# Step 3: Input text
texts = [
    "I am learning Hugging Face",
    "AutoTokenizer converts text into numbers"
]

# Step 4: Tokenization
inputs = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

# Step 5: Display generated fields
print("Input IDs:")
print(inputs["input_ids"])

print("Attention Mask:")
print(inputs["attention_mask"])

# Step 6: Pass inputs to model
outputs = model(**inputs)

print("Model Output Shape:")
print(outputs.last_hidden_state.shape)
```
