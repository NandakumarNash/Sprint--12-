# 01. What is Transformers?

**Transformers** is an open-source Python library developed by Hugging Face.

It provides pre-trained models that can perform different AI tasks such as:

- Text classification
- Sentiment analysis
- Text generation
- Question answering
- Text summarization
- Image classification
- Speech recognition

Instead of building and training a model from scratch, we can use an existing pre-trained model through the Transformers library.

**Example:**

We can use a pre-trained Transformer model to identify whether a customer review is positive or negative.

**Simple understanding:**

Transformers is a library that helps developers load, use, and work with pre-trained AI models easily.

# 02. Why Transformers is Used?

The Transformers library is used because it makes working with pre-trained AI models easier and faster.

**Main reasons:**

1. **Pre-trained Models:** We can use models that are already trained on large datasets.
2. **Easy Model Loading:** We can load models using simple Python commands.
3. **Multiple AI Tasks:** It supports NLP, computer vision, and audio tasks.
4. **Less Training Time:** We do not need to train every model from scratch.
5. **Transfer Learning:** We can fine-tune existing models for specific tasks.
6. **Easy Integration:** It works with frameworks such as PyTorch and TensorFlow.
7. **Model Reusability:** We can use different pre-trained models for different applications.

**Example:**

Suppose we want to build a sentiment analysis application.

Without Transformers:
- We may need to build and train a model.
- We need to prepare the training process.
- We need to evaluate the model.

With Transformers:
- We can load an existing sentiment analysis model.
- Provide customer reviews as input.
- Get positive or negative predictions.

**Conclusion:**

Transformers reduces development effort by providing reusable pre-trained models and simple tools to perform AI tasks.

# 03. Model Classes

**What are Model Classes?**

Model classes are Python classes provided by the Transformers library that allow us to load and use pre-trained AI models.

Each model class is designed to perform a particular type of AI task.

**Examples of Model Classes:**

1. `BertModel` – Loads the BERT model and provides contextual text representations.
2. `BertForSequenceClassification` – Performs text classification.
3. `GPT2LMHeadModel` – Performs text generation.
4. `ViTForImageClassification` – Performs image classification.
5. `WhisperForConditionalGeneration` – Performs speech-related generation tasks, including transcription.

**How Model Classes Work:**

1. Load a pre-trained model.
2. Provide the required input.
3. The model processes the input.
4. Get the model output.

**Simple Understanding:**

Model classes are responsible for processing input data and generating predictions or outputs using learned model weights.

**Important:**

Different model classes support different tasks and input formats.

```python
from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

result = classifier("I love learning Artificial Intelligence.")

print(result)
```

# 04. Tokenizer Classes

**What are Tokenizer Classes?**

Tokenizer classes convert human-readable text into numerical tokens that a Transformer model can understand.

Transformer models cannot directly process normal text. They require numerical input.

**Main responsibilities:**

1. Convert text into tokens.
2. Convert tokens into token IDs.
3. Generate attention masks.
4. Prepare text data for model input.
5. Convert token IDs back into readable text when required.

**Example:**

Input text:

`I love Artificial Intelligence`

Tokenization:

`["I", "love", "Artificial", "Intelligence"]`

Token IDs:

`[101, 1045, 2293, 9554, 4454, 102]`

*Note: Token IDs depend on the tokenizer. The example IDs are illustrative.*

**Common Tokenizer Classes:**

- `BertTokenizer` – Tokenizer for BERT models.
- `GPT2Tokenizer` – Tokenizer for GPT-2 models.
- `RobertaTokenizer` – Tokenizer for RoBERTa models.
- `AutoTokenizer` – Automatically selects the correct tokenizer.

**Simple Understanding:**

Tokenizer classes convert text into numerical information so that Transformer models can process it.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "distilbert/distilbert-base-uncased"
)

text = "I love Artificial Intelligence"

tokens = tokenizer.tokenize(text)

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

# 05. Configuration Classes

**What are Configuration Classes?**

Configuration classes store the settings and architectural details required by a Transformer model.

They describe how a model is structured and how it should be initialized.

**Common configuration information:**

1. `hidden_size` – Size of the model's hidden representations.
2. `num_hidden_layers` – Number of Transformer layers.
3. `num_attention_heads` – Number of attention heads.
4. `vocab_size` – Number of tokens in the vocabulary.
5. `max_position_embeddings` – Maximum supported position embeddings.

**Common Configuration Classes:**

- `BertConfig` – Configuration for BERT models.
- `GPT2Config` – Configuration for GPT-2 models.
- `RobertaConfig` – Configuration for RoBERTa models.
- `AutoConfig` – Automatically loads the appropriate model configuration.

**Example:**

A BERT configuration describes the model architecture, including its number of layers, hidden size, and attention heads.

**Simple Understanding:**

Configuration classes describe the model's architecture and settings. They do not contain the actual trained model weights.

```python
from transformers import AutoConfig

config = AutoConfig.from_pretrained(
    "distilbert/distilbert-base-uncased"
)

print("Model Type:", config.model_type)
print("Hidden Size:", config.dim)
print("Number of Layers:", config.n_layers)
print("Attention Heads:", config.n_heads)
print("Vocabulary Size:", config.vocab_size)
```

# 06. Auto Classes

**What are Auto Classes?**

Auto Classes are special classes in the Hugging Face Transformers library that automatically identify and load the appropriate model, tokenizer, or configuration based on a model repository.

Instead of manually selecting a specific model class, we can use Auto Classes.

**Common Auto Classes:**

1. `AutoTokenizer` – Automatically loads the correct tokenizer.
2. `AutoModel` – Loads the base model architecture and weights.
3. `AutoConfig` – Loads the model configuration.
4. `AutoModelForSequenceClassification` – Loads a model for text classification.
5. `AutoModelForCausalLM` – Loads a model for text generation.
6. `AutoModelForImageClassification` – Loads a model for image classification.

**Example:**

If we want to use a BERT model, we do not need to manually select `BertModel` or `BertTokenizer`.

We can use:

- `AutoTokenizer` to load its tokenizer.
- `AutoModel` to load its model.

The library automatically identifies the correct architecture.

**Simple Understanding:**

Auto Classes make model loading easier by automatically selecting the appropriate class based on the model repository and task.

```python
from transformers import AutoTokenizer, AutoModel

model_name = "distilbert/distilbert-base-uncased"

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load model
model = AutoModel.from_pretrained(model_name)

print("Tokenizer:", type(tokenizer).__name__)
print("Model:", type(model).__name__)
```

# 07. Pipelines

**What are Pipelines?**

Pipelines are high-level tools provided by the Hugging Face Transformers library to perform AI tasks with minimal Python code.

They simplify the process of:
- Loading a pre-trained model.
- Loading its tokenizer or processor.
- Preparing input data.
- Running model inference.
- Producing readable output.

**Common Pipeline Tasks:**

1. `sentiment-analysis` – Identifies positive or negative sentiment.
2. `text-generation` – Generates text.
3. `summarization` – Summarizes long text.
4. `question-answering` – Answers questions based on given context.
5. `translation` – Translates text between languages.
6. `image-classification` – Identifies objects or categories in images.
7. `automatic-speech-recognition` – Converts speech into text.

**Example:**

We want to identify the sentiment of a customer review.

Input:

`The product is excellent. I really like it.`

Pipeline output:

`POSITIVE`

We do not need to manually load the model and tokenizer separately.

**Simple Understanding:**

Pipelines provide an easy way to perform different AI tasks using pre-trained models without writing the complete model-loading and inference process manually.

```python
from transformers import pipeline

sentiment_pipeline = pipeline("sentiment-analysis")

result = sentiment_pipeline(
    "The product is excellent. I really like it."
)

print(result)
```

# Task 1 – Sentiment Analysis

**What is Sentiment Analysis?**

Sentiment Analysis is an NLP task that identifies the emotional tone of text.

It classifies text into categories such as:
- Positive
- Negative
- Neutral (depending on the model)

**Model Used:**

`distilbert/distilbert-base-uncased-finetuned-sst-2-english`

**Input:**

`I really enjoyed learning Artificial Intelligence.`

**Expected Output:**

- Label: POSITIVE
- Score: High confidence

**Understanding:**

The pre-trained model reads the input sentence and predicts its sentiment without requiring us to train a new model.

```python
from transformers import pipeline

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

text = "I really enjoyed learning Artificial Intelligence."

result = sentiment_pipeline(text)

print(result)
```

# Task 2 – Text Generation

**What is Text Generation?**

Text Generation is an NLP task where an AI model generates new text based on a given input prompt.

**Model Used:**

`openai-community/gpt2`

**Input Prompt:**

`Artificial Intelligence is transforming the world because`

**Expected Output:**

The model generates additional words and sentences based on the input prompt.

**Understanding:**

GPT-2 uses the previous words as context to predict the next words and generate meaningful text.

**Important:**

The generated output may vary between executions. It is not guaranteed to be factually correct.

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

prompt = "Artificial Intelligence is transforming the world because"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```

# Task 3 – Question Answering

**What is Question Answering?**

Question Answering is an NLP task where a pre-trained model finds an answer to a question using the given context.

**Model Used:**

`distilbert/distilbert-base-cased-distilled-squad`

**Context:**

`Hugging Face is a company that develops tools for building and using machine learning models. Its Transformers library provides access to pre-trained models for NLP, computer vision, and audio tasks.`

**Question:**

`What does the Transformers library provide?`

**Expected Answer:**

`Access to pre-trained models for NLP, computer vision, and audio tasks.`

**Understanding:**

The model reads the context and question, identifies the relevant information, and extracts the answer from the context.

```python
pip install "transformers>=4.45,<5"
```

```python
from transformers import pipeline

qa_pipeline = pipeline(
    "question-answering",
    model="distilbert/distilbert-base-cased-distilled-squad"
)

context = """
Hugging Face is a company that develops tools for building
and using machine learning models. Its Transformers library
provides access to pre-trained models for NLP,
computer vision, and audio tasks.
"""

question = "What does the Transformers library provide?"

result = qa_pipeline(
    question=question,
    context=context
)

print(result)
print("Answer:", result["answer"])
```
