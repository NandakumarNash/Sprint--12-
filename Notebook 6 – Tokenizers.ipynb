1. What is a Token?
A token is a small unit of text produced by a tokenizer.

A token can be:
- A complete word.
- A part of a word.
- A punctuation mark.
- A special token.

```python
text = "I love Machine Learning"
```

2. What is a Tokenizer?

A tokenizer is a tool that converts normal text into tokens and then converts those tokens into numerical IDs.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

text = "I love Machine Learning"

tokens = tokenizer.tokenize(text)

print(tokens)
```

4. What are Token IDs?

Token IDs are numerical representations of tokens.

```python
text = "I love Machine Learning"

encoded = tokenizer(text)

print(encoded["input_ids"])
```

5. What are Special Tokens?
Special tokens are predefined tokens used to control or structure model input.

For BERT, important special tokens include:
Token	Purpose

[CLS]	Marks the beginning of a sequence

[SEP]	Separates or marks the end of a sequence

[PAD]	Fills unused positions

[UNK]	Represents an unknown token

print(tokenizer.cls_token)
print(tokenizer.sep_token)
print(tokenizer.pad_token)
print(tokenizer.unk_token)

6. Understanding Attention Mask

An attention mask tells the model which positions contain actual tokens and which positions are padding.
- 1 = Actual token.
- 0 = Padding token.

```python
text = [
    "I love AI",
    "I love Artificial Intelligence and Machine Learning"
]

encoded = tokenizer(
    text,
    padding=True
)

print(encoded["input_ids"])
print(encoded["attention_mask"])
```

7. What is Padding?

Padding adds special [PAD] tokens to shorter sentences so that multiple sentences have equal lengths.

Sentence 1: I love AI

Sentence 2: I love Machine Learning

After padding:

Sentence 1: I love AI [PAD] [PAD]

Sentence 2: I love Machine Learning

```python
sentences = [
    "I love AI",
    "I love Machine Learning"
]

encoded = tokenizer(
    sentences,
    padding=True
)

print(encoded["input_ids"])
```

8. What is Truncation?

Truncation removes tokens when the input exceeds the specified maximum length.

```python
Original:
I love learning Artificial Intelligence and Deep Learning

After truncation:
I love learning Artificial Intelligence
```

```python
text = "I love learning Artificial Intelligence and Deep Learning"

encoded = tokenizer(
    text,
    truncation=True,
    max_length=6
)

print(encoded["input_ids"])
```

9. What is Maximum Sequence Length?

Maximum sequence length is the maximum number of tokens allowed in a model input.

```python
encoded = tokenizer(
    "I love learning Artificial Intelligence",
    max_length=8,
    padding="max_length",
    truncation=True
)

print(len(encoded["input_ids"]))
```

Here:
- max_length=8 sets the maximum length.
- padding="max_length" fills the input to 8 positions.
- truncation=True removes excess tokens if necessary.
Remember: Special tokens also occupy positions in the sequence.

```python
sentences = [
    "I love AI",
    "Machine Learning is interesting",
    "I am learning Artificial Intelligence and Deep Learning"
]

for sentence in sentences:

    tokens = tokenizer.tokenize(sentence)

    print("Sentence:", sentence)
    print("Tokens:", tokens)
    print("Token Count:", len(tokens))
    print()
```
