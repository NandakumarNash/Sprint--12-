# 03 – Model Repository Analysis

## 1. What is a Model Repository?

A model repository is a collection of files stored together on Hugging Face Hub.

These files contain everything required to understand, load, and use a pre-trained model.

A model repository usually contains:

- Model configuration files
- Model weight files
- Tokenizer files
- Vocabulary files
- README / Model Card

Each file has a different responsibility.

**Example:**

When we download a pre-trained text classification model, its repository may contain:

- `config.json` – Defines the model architecture and configuration.
- `model.safetensors` – Stores the trained model weights.
- `tokenizer.json` – Stores tokenizer information.
- `vocab.txt` – Contains vocabulary tokens.
- `README.md` – Explains the model and its usage.

**Important:** These files work together. Configuration tells the software how to construct the model, weights provide its learned knowledge, and tokenizer files help convert input text into a format the model understands.

```python

from huggingface_hub import list_repo_files

model_name = "google-bert/bert-base-uncased"

files = list_repo_files(model_name)

for file in files:
    print(file)
```

## 2. Model Configuration – config.json

`config.json` is an important file in a Hugging Face model repository.

It contains information about the model architecture and its configuration.

It tells the Transformers library how to create the model structure before loading the trained weights.

**Common information stored in config.json:**

1. `model_type` – Identifies the model architecture, such as BERT.
2. `hidden_size` – Defines the size of the model's hidden representations.
3. `num_hidden_layers` – Specifies the number of transformer layers.
4. `num_attention_heads` – Specifies the number of attention heads.
5. `vocab_size` – Defines the vocabulary size.
6. `num_labels` – Specifies the number of output classes for classification models.

**Why is config.json important?**

Without the correct configuration, the software may not know how to construct the model architecture correctly.

The configuration and model weights must be compatible with each other.

**Example:**

For a BERT model, config.json tells the software how many transformer layers and attention heads to create. The model weight file contains the learned values for those layers.

```python
from transformers import AutoConfig

model_name = "google-bert/bert-base-uncased"

config = AutoConfig.from_pretrained(model_name)

print("Model Type:", config.model_type)
print("Hidden Size:", config.hidden_size)
print("Number of Layers:", config.num_hidden_layers)
print("Attention Heads:", config.num_attention_heads)
print("Vocabulary Size:", config.vocab_size)
```

## 3. Model Weight Files

Model weight files store the values learned by a model during training.

During training, a model learns patterns and relationships from data. These learned values are called **model weights**.

When we use a pre-trained model, the saved weights are loaded into the model architecture to perform predictions or inference.

**Common model weight file formats:**

1. `model.safetensors` – Stores model weights in the SafeTensors format.
2. `pytorch_model.bin` – Stores model weights in PyTorch format.
3. `tf_model.h5` – Stores model weights in TensorFlow format.
4. `model.safetensors.index.json` – Contains weight-to-file mapping information when weights are split across multiple files.

**Why are model weight files important?**

Without trained weights, the model has no previously learned knowledge to perform its intended task.

The configuration defines the model structure, while the weight files provide the learned parameters.

**Example:**

A pre-trained BERT model contains learned weights from its training process. When we load these weights into the BERT architecture, the model can process text and generate meaningful outputs.

```python
from transformers import AutoModel

model_name = "google-bert/bert-base-uncased"

model = AutoModel.from_pretrained(model_name)

print("Model loaded successfully!")

print("Total parameters:", sum(
    p.numel() for p in model.parameters()
))
```

What happens here?
1. Downloads the pre-trained model files if they are not cached.
2. Reads the model configuration.
3. Loads the trained weights.
4. Creates a ready-to-use model.
Important distinction:
File	Purpose
config.json	Defines model architecture
model.safetensors	Stores trained weights

## 4. Tokenizer Files

A tokenizer converts human-readable text into tokens and then into numerical token IDs that a model can understand.

Tokenizer files store the information required to perform this conversion.

**Common tokenizer files:**

1. `tokenizer.json` – Stores the tokenizer configuration, vocabulary and tokenization rules in a combined format.
2. `tokenizer_config.json` – Stores tokenizer settings, such as special token configuration and tokenizer-related options.
3. `special_tokens_map.json` – Maps special tokens such as `[CLS]`, `[SEP]`, `[PAD]` and `[UNK]`.

**Why are tokenizer files important?**

A language model cannot directly understand raw human language. It requires numerical input.

The tokenizer converts text into token IDs before sending it to the model.

**Example:**

Input text:

`I love machine learning`

The tokenizer converts the text into tokens and then into numerical IDs.

These IDs are passed to the model as input.

**Important:** The tokenizer and model must be compatible. Using the wrong tokenizer can produce incorrect input representations.

```python
from transformers import AutoTokenizer

model_name = "google-bert/bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

text = "I love machine learning"

tokens = tokenizer.tokenize(text)

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Original Text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

## 5. Vocabulary Files

A vocabulary file contains the collection of tokens that a tokenizer recognizes.

Each token is assigned a unique numerical ID. These IDs are used to convert text into numerical input for a language model.

**Common vocabulary files:**

1. `vocab.txt` – Contains vocabulary tokens, commonly used by BERT-based models.
2. `vocab.json` – Stores vocabulary tokens and their corresponding numerical IDs, commonly used by some other tokenizer architectures.
3. `merges.txt` – Contains merge rules used by Byte Pair Encoding (BPE) tokenizers.

**Why are vocabulary files important?**

The vocabulary determines which tokens the tokenizer can recognize and convert into IDs.

If a word is not available as a complete token, the tokenizer may split it into smaller subword tokens.

**Example:**

A BERT vocabulary file may contain:

- hello
- machine
- learning
- computer

Each token has its own numerical ID.

When the tokenizer processes text, it uses the vocabulary to find the corresponding token IDs.

**Important:** Vocabulary files are not the same as model weights. Vocabulary files support text processing, while model weights store learned parameters.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-uncased"
)

vocabulary = tokenizer.get_vocab()

print("Vocabulary Size:", len(vocabulary))

for token, token_id in list(vocabulary.items())[:10]:
    print(token, ":", token_id)
```

## 6. README.md / Model Card

`README.md` is the documentation file of a Hugging Face model repository.

On Hugging Face Hub, this file is commonly called a **Model Card**.

It explains what the model does, how it was developed, and how it should be used.

**Common information in a Model Card:**

1. Model Description – Explains the purpose of the model.
2. Model Architecture – Describes the architecture used.
3. Intended Uses – Explains the tasks the model is designed for.
4. Training Information – Provides details about training data and methods.
5. Evaluation – Describes model performance and evaluation results.
6. Limitations – Explains the model's weaknesses and restrictions.
7. Usage Examples – Shows how to load and use the model.

**Why is a Model Card important?**

Before selecting a pre-trained model, we need to understand whether it is suitable for our task.

The Model Card helps us understand the model's capabilities, intended applications, limitations, and training background.

**Example:**

If we want to perform sentiment analysis, the Model Card helps us check whether the model supports sentiment classification.

It also helps us understand its training data, supported languages, and limitations.

**Important:** A Model Card provides documentation about a model. It does not contain the model's trained weights.

```python
from huggingface_hub import hf_hub_download

model_name = "google-bert/bert-base-uncased"

readme_path = hf_hub_download(
    repo_id=model_name,
    filename="README.md"
)

with open(readme_path, "r", encoding="utf-8") as file:
    readme_content = file.read()

print(readme_content[:2000])
```

## 7. Inspect Three Model Repositories

We selected three pre-trained models from Hugging Face Hub to understand how model repositories are organized.

**Repository 1: BERT**

Model: `google-bert/bert-base-uncased`

BERT is a transformer-based language model designed to understand text.

**Repository 2: DistilBERT**

Model: `distilbert/distilbert-base-uncased`

DistilBERT is a smaller and faster version of BERT. It uses knowledge distillation to reduce model size while retaining much of BERT's capabilities.

**Repository 3: GPT-2**

Model: `openai-community/gpt2`

GPT-2 is a transformer-based language model designed for text generation.

**Purpose of this inspection:**

- Identify configuration files.
- Identify model weight files.
- Identify tokenizer files.
- Identify vocabulary files.
- Identify README / Model Card.
- Understand how repository structures differ between models.

```python
from huggingface_hub import list_repo_files

models = [
    "google-bert/bert-base-uncased",
    "distilbert/distilbert-base-uncased",
    "openai-community/gpt2"
]

for model in models:
    print("\nModel Repository:", model)
    print("-" * 50)

    files = list_repo_files(model)

    for file in files:
        print(file)
```

**Observation:**

Different models may contain different repository files depending on their architecture and tokenizer.

Some repositories combine tokenizer information into one file, while others store vocabulary and tokenizer rules separately.

Model weights may also be stored in one file or divided into multiple files.

Therefore, every model repository does not necessarily have the same structure.

## 8. Relationship Between Model Repository Files

A model repository contains multiple files that work together to make a pre-trained model usable.

Each file has a specific responsibility.

**1. Model Configuration (`config.json`)**

Defines the model architecture, number of layers, hidden size, and other settings.

**2. Model Weights (`model.safetensors` / `pytorch_model.bin`)**

Stores the learned parameters obtained during model training.

**3. Tokenizer Files (`tokenizer.json`, `tokenizer_config.json`)**

Define how input text is processed and converted into tokens and numerical IDs.

**4. Vocabulary Files (`vocab.txt`, `vocab.json`)**

Contain the available tokens and their corresponding IDs.

**5. Model Card (`README.md`)**

Provides information about the model, its purpose, training, usage, and limitations.

### How do they work together?

1. The Model Card helps users understand and select a suitable model.
2. The tokenizer processes the input text.
3. Vocabulary information helps convert tokens into numerical IDs.
4. The configuration defines the model architecture.
5. The model weights are loaded into that architecture.
6. The model processes the numerical input and produces an output.

**Conclusion:**

A model repository is not just a collection of files. It is an organized package containing the configuration, learned parameters, input-processing resources, and documentation needed to understand and use a pre-trained model.
